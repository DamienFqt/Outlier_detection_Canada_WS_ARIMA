## Libraries

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
```

## Data

```python
# chemin du notebook
notebook_path = Path().resolve()
# chemin du CSV relatif à la racine du projet
data_path = notebook_path.parent / "data" / "processed.csv"
w = pd.read_csv(data_path)
w
```

# Descriptive statistics

```python
# Global descriptive statistics
w['MEAN_TEMPERATURE_MONTREAL'].describe()
```

```python
# Set 'date' as the index to make it a time series
w.set_index('date', inplace=True)

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(w['MEAN_TEMPERATURE_MONTREAL'], label='Mean Temperature (Montréal)', color='tab:blue')
plt.title('Mean Temperature in Montréal (2000-2019)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

w.reset_index(inplace=True)
```

```python
# Filter the data for the years 2017, 2018, and 2019
w3 = w[(w['year'] == 2017) | (w['year'] == 2018) | (w['year'] == 2019)]

# Set 'date' as the index to make it a time series
w3.set_index('date', inplace=True)

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(w3['MEAN_TEMPERATURE_MONTREAL'], label='Mean Temperature (Montréal)', color='tab:blue')
plt.title('Mean Temperature in Montréal (2017-2019)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

w3.reset_index(inplace=True)
```

```python
# Filter the data for the year 2019
w1 = w[(w['year'] == 2019)]

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(w1['MEAN_TEMPERATURE_MONTREAL'], label='Mean Temperature (Montréal)', color='tab:blue')
plt.title('Evolution des températures quotidiennes moyennes à Montréal en 2019')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
